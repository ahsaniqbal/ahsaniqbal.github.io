For per-channel symmetric quantization:

$$
s_c = \frac{\max_j |W_{c,j}|}{127}.
$$


```python
import torch

weights = torch.randn(4, 16)
abs_max = weights.abs().max(dim=-1).values
scales = abs_max / 127
q = torch.round(weights / scales[:, None]).clamp(-127, 127).to(torch.int8)

q.shape, scales
```

## Reconstruction error

The dequantized approximation is

$$\hat W_{c,j} = s_c q_{c,j}.$$


```python
dequantized = q.float() * scales[:, None]
mae = (weights - dequantized).abs().mean()
mae
```
